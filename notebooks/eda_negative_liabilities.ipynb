```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/hospital_cost_reports.db")
d = pd.read_sql("""
    SELECT provider_ccn, file_year, ccn_facility_type, type_of_control,
           total_assets, total_liabilities, total_fund_balances,
           total_liabilities_and_fund_balances
    FROM cost_report_clean
    WHERE total_liabilities < 0 AND ccn_facility_type IN ('STH', 'CAH')
""", conn)

print("Rows with negative total_liabilities:", len(d))
print("Hospitals:", d.provider_ccn.nunique())
print(d.ccn_facility_type.value_counts())

# Does assets = liabilities + fund balances still hold on these rows?
gap = d.total_assets - (d.total_liabilities + d.total_fund_balances)
print("Rows where the balance sheet balances (within $1):", (gap.abs() <= 1).sum())
d.head(10)
```
